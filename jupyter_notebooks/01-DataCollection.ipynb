We drag and drop the Kaggle.json authentication token into the workspace


Then we run the cell, so that the token is recognized in the session

```python
os.environ['KAGGLE_CONFIG_DIR'] = os.getcwd()
if os.name == "posix":
    os.chmod("kaggle.json", 0o600)
```

Define the Kaggle dataset, and destination folder and download it.

```python
KaggleDatasetPath = "codeinstitute/housing-prices-data"
DestinationFolder = "inputs/datasets/raw"

if os.path.exists(f"{DestinationFolder}/housing-prices-data.zip") or os.path.exists(
    f"{DestinationFolder}/house-price-20211124T154130Z-001"
):
    print("Dataset already present locally, skipping download.")
else:
    ! kaggle datasets download -d {KaggleDatasetPath} -p {DestinationFolder}
```

The data cell is in a zip file. We Unzip the downloaded file, delete the zip file and delete the kaggle token file

```python
import zipfile

zip_path = f"{DestinationFolder}/housing-prices-data.zip"
extracted_marker = f"{DestinationFolder}/house-price-20211124T154130Z-001"

if os.path.exists(extracted_marker):
    print("Already extracted, skipping.")
elif os.path.exists(zip_path):
    with zipfile.ZipFile(zip_path) as zip_ref:
        zip_ref.extractall(DestinationFolder)
    os.remove(zip_path)
    print("Extracted and removed zip file.")
else:
    print("No zip file found to extract.")
```
